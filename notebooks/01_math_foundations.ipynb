## Eigen Vectors and  Values
An Eigen Vector is a special vector that stays in it's span (set of all possible linear combinations you can create while scaling those vectors and adding them together) after a linear transformation. It is either stretched or squished by a factor that we call the Eigen value. It is used in principal analysis (PCA) and facial recognition algorithms. I breaks a square matrix down into its fundamental structural components.

```python
eigenvalues, eigenvectors = np.linalg.eig(Mat_A)
print("Eigenvalues: ", eigenvalues)
print("Eigenvectors: \n", eigenvectors)
```

## SVD (Singular value decomposition)
SVD factorizes any matrix A (it doesn't have to be square) into three distinct  matrices:
    \(\mathbf{A}=\mathbf{U}\cdot \mathbf{\Sigma }\cdot \mathbf{V}^{T}\)
- U: Left singular vectors (orthogonal matrix).
- Σ: Singular values (represented by NumPy as a flat 1D vector of values sorted from largest to smallest).
- \(V^{T}\): Right singular vectors (orthogonal matrix, already transposed by NumPy).
- SVD is used in dimentionality reduction

```python
rng = np.random.default_rng(seed=75)   #generates a random Matrix
Mat_C = rng.integers(low=0, high=100, size=(4,3))     # produces 4x3 matrix with integers btw 0 & 99
print("Matrix B:\n", Mat_C)
print("Shape: ", Mat_C.shape)
```

```python
U, s, Vt = np.linalg.svd(Mat_C)

print("--- Matrix U --- \n", U)
print("\n--- Singular Values (s) --- \n", s)
print("\n--- Matrix V^T --- \n", Vt)
```

# Section B : Calculus: Derivatives and Gradients

## Derivative

```python

# 1. Define the mathematical function and it's derivative
def f(x):
    return x**2
def f_prime(x):
    return 2*x

# 2. Generate a dense range of x values (e.g., from -5 to 5, with 500 points)
x = np.linspace(-5, 5, 500)

# 3. Calculate corresponding y values
y = f(x)
y_prime = f_prime(x)

# 4. Create the plot and customize it
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, label=r'$f(x) = x^2$', color='royalblue', linewidth=2)
ax.plot(x, y_prime, label=r'$f(x) = 2x$', color='crimson', linewidth=2)

# 5. Add labels, grid, and legend
ax.set_title('Plot of a Quadratic Function', fontsize=14)
ax.set_xlabel('x axis', fontsize=12)
ax.set_ylabel('y axis', fontsize=12)
#ax.axhline(0, color='black', linewidth=0.5, linestyle='--')  # X-axis line
#ax.axvline(0, color='black', linewidth=0.5, linestyle='--')  # Y-axis line
#ax.grid(True, linestyle=':', alpha=0.6)
#fig.legend(loc='outside upper right')
fig.legend(loc='outside lower right')

# 6. Display the plot
plt.show()
```

```python
def numerical_derivative(f, x, h = 1e-5):
    '''
    aproximating the gradient of a function f at a point x
    using the symetric central difference method.
    '''
    y_ahead = f(x + h)
    y_behind = f(x - h)

    slope = (y_ahead - y_behind)/(2*h)
    return slope

def my_curve(x):
    return x**3

approximate_derivative = numerical_derivative(my_curve,4)
print(f"approximate slope = {approximate_derivative}")
```

direction in which the loss decreases fastest.

# Section C : Probability & Statistics

```python
df = pd.read_csv('/media/marielle-daha/52f496b1-0f18-44d6-b9c5-4fece19d0001/marielle-daha/retainiq-churn/data/WA_Fn-UseC_-Telco-Customer-Churn.csv')
#loading the dataset
```

## Measure of central tendency 

```python
data = df['MonthlyCharges']

## Mean
'''
The average value calculated by adding all 
the data points and dividing by the total count
'''
mean_val = data.mean()

## Median
'''
the middle value when all numbers are sorted in asscending order
'''
median_val = data.median()

## standard deviation 
'''
A measure of how spread out the data points are from the mean
'''
standard_deviation = data.std()

## Skewness
'''
A measure of the asymmetry of the data distribution. Using Karl Pearson's second skewness coefficient, it can be approximated as: 
Skewness = (3 * (Mean - Median)) / Standard Deviation
'''
skewness_val = data.skew()

print(f'The mean value for the monthly charges is : {mean_val}')
print(f'The median value for the monthly charges is : {median_val}')
print(f'The standard deviation value for the monthly charges is : {standard_deviation}')
print(f'The skewness value for the monthly charges is : {skewness_val}')
```

## Histogram

```python
fig, ax = plt.subplots(tight_layout=True)

ax.hist(
    data,              # the values to plot (e.g. the Monthly_Charges column)
    bins=30,           # split the data range into 30 equal-width intervals -> 30 bars
    density=True,      # rescale bars so total area = 1 (same scale as norm.pdf); y-axis becomes density, not counts
    alpha=0.6,         # transparency (0 = invisible, 1 = solid) so the curve stands out
    label="Data"       # name shown in the legend for these bars
)

x = np.linspace(data.min(), data.max(), 500)
ax.plot(x, norm.pdf(x, mean_val, standard_deviation), color="crimson", linewidth=2,
        label=f"Normal (μ={mean_val:.2f}, σ={standard_deviation:.2f})")

ax.set_title('Histogram of Monthly_Charges')
ax.set_xlabel('Values')
ax.set_ylabel('Density')   
ax.legend()

plt.show()
```

## Probability

```python
# Total number of records in the dataset
total_customers = len(df)

# Individual probabilities (P(A) and P(B))
p_churn = len(df[df['Churn'] == 'Yes']) / total_customers
p_month_to_month = len(df[df['Contract'] == 'Month-to-month']) / total_customers

# Likelihood (P(B|A))
# Of those who churned, how many had a month-to-month contract?
p_month_to_month_given_churn = len(df[(df['Churn'] == 'Yes') & (df['Contract'] == 'Month-to-month')]) / len(df[df['Churn'] == 'Yes'])

# Apply Bayes' Theorem
bayes_prob = (p_month_to_month_given_churn * p_churn) / p_month_to_month

print(f"Probability via Bayes' Theorem: {bayes_prob:.4f}")
```
